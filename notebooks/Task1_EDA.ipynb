```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.columns = df.columns.str.lower()
```

```python
categorical = df.select_dtypes(include="object").columns

for col in categorical:
    df[col] = df[col].astype("category")
```

```python
Q1 = df["monthlycharges"].quantile(.25)

Q3 = df["monthlycharges"].quantile(.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers = df[(df["monthlycharges"]<lower) | (df["monthlycharges"]>upper)]

print(outliers.shape)
```

✔ Converted TotalCharges to numeric

✔ Filled missing values with median

✔ Removed duplicate rows

✔ Converted categorical columns

✔ Standardized column names

✔ Checked outliers using IQR

Key Findings

1. Most customers did not churn.

2. Customers with month-to-month contracts showed higher churn.

3. Customers paying higher monthly charges were more likely to churn.

4. Longer tenure customers had lower churn.

5. Monthly charges have a wider spread than tenure.
