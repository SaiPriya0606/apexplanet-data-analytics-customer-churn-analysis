```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px

plt.style.use("ggplot")
```

```python
df = pd.read_csv("../data/processed/telco_cleaned.csv")
```

```python
numeric = df.select_dtypes(include=np.number)

plt.figure(figsize=(8,6))

sns.heatmap(numeric.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.savefig("../visualizations/png/heatmap.png")

plt.show()
```

```python
sns.pairplot(df[["tenure","monthlycharges","totalcharges"]])

plt.savefig("../visualizations/png/pairplot.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxenplot(x="churn", y="monthlycharges", data=df)

plt.savefig("../visualizations/png/boxen.png")

plt.show()
```

```python
g = sns.FacetGrid(df, col="gender", hue="churn")

g.map(sns.scatterplot, "tenure", "monthlycharges")

g.add_legend()

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="churn")

plt.savefig("../visualizations/png/churn_count.png")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="contract")

plt.xticks(rotation=20)

plt.savefig("../visualizations/png/contracts.png")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="paymentmethod")

plt.xticks(rotation=40)

plt.savefig("../visualizations/png/payment.png")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="internetservice")

plt.savefig("../visualizations/png/internet.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["monthlycharges"])

plt.savefig("../visualizations/png/histogram.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x="tenure",
                y="monthlycharges",
                hue="churn")

plt.savefig("../visualizations/png/scatter.png")

plt.show()
```

```python
df["churn"].value_counts().plot.pie(autopct="%1.1f%%")

plt.savefig("../visualizations/png/pie.png")

plt.show()
```

```python
sns.violinplot(x="churn",
               y="monthlycharges",
               data=df)

plt.savefig("../visualizations/png/violin.png")

plt.show()
```

```python
sns.kdeplot(df["monthlycharges"], fill=True)

plt.savefig("../visualizations/png/kde.png")

plt.show()
```

```python
sns.histplot(df["tenure"])

plt.savefig("../visualizations/png/tenure.png")

plt.show()
```

```python
sns.boxplot(y=df["monthlycharges"])

plt.savefig("../visualizations/png/monthlycharges.png")

plt.show()
```

```python
fig = px.scatter(df,
                 x="tenure",
                 y="monthlycharges",
                 color="churn")

fig.write_html("../visualizations/html/scatter.html")

fig.show()
```

```python
fig = px.bar(df,
             x="contract",
             color="churn")

fig.write_html("../visualizations/html/bar.html")

fig.show()
```

```python
fig = px.pie(df,
             names="internetservice")

fig.write_html("../visualizations/html/pie.html")

fig.show()
```

```python
fig = px.histogram(df,
                   x="monthlycharges")

fig.write_html("../visualizations/html/histogram.html")

fig.show()
```
