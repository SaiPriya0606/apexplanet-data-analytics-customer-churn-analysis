```python
import pandas as pd
import sqlite3

from sqlalchemy import create_engine
```

```python
df = pd.read_csv("../data/processed/telco_cleaned.csv")
```

```python
conn = sqlite3.connect("../database/telco.db")
print("database connected successfully")
```

```python
df.to_sql("customers", conn, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM customers LIMIT 5", conn)
```

```python
query = """
SELECT contract,
AVG(monthlycharges)
FROM customers
GROUP BY contract;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT churn,
COUNT(*)
FROM customers
GROUP BY churn;
"""

result = pd.read_sql(query, conn)

result
```

```python
contract = "Month-to-month"

query = """
SELECT *
FROM customers
WHERE contract=?
"""

pd.read_sql(query, conn, params=(contract,))
```

```python
query = """
SELECT customerid,
monthlycharges,

ROW_NUMBER()
OVER(ORDER BY monthlycharges DESC)
AS Rank

FROM customers;
"""

result = pd.read_sql(query, conn)

result
```

```python
query = """
SELECT customerid,
monthlycharges,

LAG(monthlycharges)
OVER(ORDER BY tenure)
AS PreviousCharge

FROM customers;
"""

result = pd.read_sql(query, conn)

result
```
